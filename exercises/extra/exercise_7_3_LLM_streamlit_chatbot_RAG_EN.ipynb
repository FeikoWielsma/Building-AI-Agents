## Exercises: Streamlit RAG chatbot

Exercise:
1. Using the instructions in `7_3_LLM_streamlit_chatbot_RAG/README.md`, run the chatbot application.
2. Replace the text files with your own in the folder `7_3_LLM_streamlit_chatbot_RAG/data` (these could be, for example, the texts of a few articles copied from Wikipedia).
3. Restart the chatbot and ask a question about the information contained in the new sources.
