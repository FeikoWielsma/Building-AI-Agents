Examples of image processing by the model.

Exercise 1 \
Load any image from disk (e.g. `sample.jpg`) and ask the model for a description in 1–2 sentences.

```python
# TODO: Load the image and build a message for the multimodal model
# HINT: You can load the image and pass it as a data URL or file-URI (depending on the client).
from base64 import b64encode
from pathlib import Path
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

img_path = Path("../../data/img/airport.jpg")  # <- replace with your own path
if img_path.exists():
    client = OpenAI()
    data = b64encode(img_path.read_bytes()).decode("utf-8")
    resp = client.responses.create(
        model="gpt-5.6-luna",
        input=[{
            "role": "user",
            "content": [
                {"type": "input_text", "text": ("Describe what's in the image in 1–2 sentences in Polish.")},
                {"type": "input_image", "image_url": f"data:image/jpeg;base64,{data}"}
            ],
        }],
    )

    response = resp.output_text
    print("Model response:", response)
else:
    print("No file - insert your own image and run again.")
```
