## Neural networks, by hand

- A network learns by nudging weights until its output matches the data
- None of this is needed to build an agent ; it is here for the vocabulary

This one is a website rather than code. We change the shape of a network and
watch it learn.

Exercise 1:
Use the TensorFlow artificial neural network simulator at https://playground.tensorflow.org and observe how the network learns patterns.
For more complex patterns (e.g. the spiral), increase the number of layers and the number of neurons in the hidden layers in order to recognize the more complicated pattern.

### Try the spiral

- Switch the dataset to the spiral and run it again with one hidden layer
- Add layers until it separates. That is the whole of "deeper network"
