## LangGraph basics

- A graph is a set of nodes reading and writing one shared state object
- A node is just a function : take the state, return the state
- No model is called here at all ; this is the plumbing on its own

Defines a state, writes a node, then wires `START -> echo -> END`.

### Exercise State + a simple node
Replace the question assigned to the "question" key and run the code below.

````python
from typing import TypedDict, Optional

class State(TypedDict, total=False):
    question: str
    answer: Optional[str]

def echo_node(state: State) -> State:
    new_state = dict(state)
    if "answer" not in new_state:
        new_state["answer"] = f"ECHO: {state.get('question','')}"
    return new_state

assert callable(echo_node)

state = State({"question": "Where has my order got to?"}) # <- replace with your own text
echo_node(state)
````

### Solution

````python
from typing import TypedDict, Optional

class State(TypedDict, total=False):
    question: str
    answer: Optional[str]

def echo_node(state: State) -> State:
    new_state = dict(state)
    if "answer" not in new_state:
        new_state["answer"] = f"ECHO: {state.get('question','')}"
    return new_state

assert callable(echo_node)

state = State({"question": "How does a StateGraph work?"})   # <- your own text
echo_node(state)
````
